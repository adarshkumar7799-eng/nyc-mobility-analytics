**Purpose**: Verification queries for each layer of the pipeline. Resusable accross the course.

**Owner**: Course Author

**Inputs**: Tables accross all three layers(bronze , Silver,gold).

**Outputs**: Validition results displayed inline, no table written.

```python
%sql
SELECT COUNT(*) AS bronze_total FROM nyc_mobility.bronze.trips;
```

```python
%sql
SELECT COUNT(*) AS bad_fares
FROM nyc_mobility.bronze.trips
WHERE fare_amount <= 0 OR fare_amount IS NULL;
```

```python
%sql
SELECT COUNT(*) AS bad_distances
FROM nyc_mobility.bronze.trips
WHERE trip_distance <= 0 OR trip_distance IS NULL;
```

```python
%sql
SELECT COUNT(*) AS bad_passenger_counts
FROM nyc_mobility.bronze.trips
WHERE passenger_count IS NULL OR passenger_count <= 0 OR passenger_count > 6;
```

```python
%sql
SELECT COUNT(*) AS bad_timestamps
FROM nyc_mobility.bronze.trips
WHERE tpep_pickup_datetime IS NULL
   OR tpep_dropoff_datetime IS NULL
   OR tpep_dropoff_datetime <= tpep_pickup_datetime;
```

```python
%sql
SELECT COUNT(*) AS rows_silver_will_drop
FROM nyc_mobility.bronze.trips
WHERE fare_amount <= 0 OR fare_amount IS NULL
   OR trip_distance <= 0 OR trip_distance IS NULL
   OR passenger_count IS NULL OR passenger_count <= 0 OR passenger_count > 6
   OR tpep_pickup_datetime IS NULL
   OR tpep_dropoff_datetime IS NULL
   OR tpep_dropoff_datetime <= tpep_pickup_datetime;
```
