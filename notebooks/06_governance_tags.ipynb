**Purpose:** Apply the Unity Catalog tag scheme (domain, layer, sensitivity, unit) across the nyc_mobility estate.

**Owner:** Course Author

**Inputs:** Existing nyc_mobility catalog, schemas, and Gold revenue columns.

**Outputs:** Classified estate: domain on the catalog, layer and sensitivity on the schemas, unit on the revenue columns.

```python
%sql
-- Catalog: the business domain
ALTER CATALOG nyc_mobility
  SET TAGS ('domain' = 'mobility');

-- Schemas: layer and sensitivity
ALTER SCHEMA nyc_mobility.bronze
  SET TAGS ('layer' = 'bronze', 'sensitivity' = 'public');
ALTER SCHEMA nyc_mobility.silver
  SET TAGS ('layer' = 'silver', 'sensitivity' = 'public');
ALTER SCHEMA nyc_mobility.gold
  SET TAGS ('layer' = 'gold', 'sensitivity' = 'public');

-- Columns: currency unit on the revenue measures
ALTER TABLE nyc_mobility.gold.revenue_by_date
  ALTER COLUMN total_revenue SET TAGS ('unit' = 'usd');
ALTER TABLE nyc_mobility.gold.trips_by_borough
  ALTER COLUMN total_revenue SET TAGS ('unit' = 'usd');
```
