**Purpose:** Query Unity Catalog lineage system tables for upstream sources and downstream impact analysis of the nyc_mobility pipeline.

**Owner:** Course Author

**Inputs:** system.access.table_lineage lineage events for the nyc_mobility estate.

**Outputs:** Read-only lineage queries: sources of silver.trips, and downstream tables affected by it.


```sql
SELECT
  source_table_full_name,
  source_table_catalog,
  source_table_schema,
  source_table_name,
  event_time
FROM system.access.table_lineage
WHERE target_table_full_name = 'nyc_mobility.silver.trips'
  AND source_table_full_name IS NOT NULL
ORDER BY event_time DESC
LIMIT 20;

```

```sql
-- All downstream dependencies of silver.trips (impact analysis)
SELECT DISTINCT
  target_table_full_name,
  target_table_catalog,
  target_table_schema,
  target_table_name
FROM system.access.table_lineage
WHERE source_table_full_name = 'nyc_mobility.silver.trips'
  AND target_table_full_name IS NOT NULL
ORDER BY target_table_full_name;
```
