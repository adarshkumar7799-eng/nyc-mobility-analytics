```sql
create or replace table nyc_mobility.bronze.taxi_zones as
select 
    cast(LocationID as int) as location_id,
    Borough,
    Zone,
    service_zone
    from read_files(
        '/Volumes/nyc_mobility/bronze/reference_data/taxi_zone_lookup.csv',
        format => 'csv',
        header => true
    );
   
```
