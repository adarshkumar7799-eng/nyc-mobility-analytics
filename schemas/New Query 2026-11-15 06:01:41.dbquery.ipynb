```sql
SELECT version, operation, operationMetrics, operationParameters
FROM (DESCRIBE HISTORY nyc_mobility.gold.revenue_by_date)
ORDER BY version DESC;



-- Query physical file metadata via hidden _metadata column
SELECT _metadata.file_name AS parquet_file, COUNT(*) AS rows_from_this_file
FROM nyc_mobility.gold.revenue_by_date
GROUP BY 1;
```
